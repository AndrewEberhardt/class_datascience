## Exercise 1: Generate data <a name="Exercise-1:-Generate-data"></a>
- As done in in the resources' notebook, generate 100 observations of CO2 emissions per capita. This variable distributes like a Normal with mean 5 tonnes and variance 4 tonnes.

```python
# set parameters: mean 5, variance 4 (so standard deviation 2)
np.random.seed(12345)
beta = 5
sigma2 = 4
sigma = np.sqrt(sigma2)
N = 100

# draw 100 observations
draws_co2 = np.random.normal(loc = beta, scale = sigma, size = N)
draws_co2
```

## Exercise 2: Density functions <a name="Exercise-2:-Density-functions"></a>

- Use a built-in function inside `stats.norm` to obtain the values of the density function (density values) for `draws_co2`, and comment what these values mean.

```python
pdf_co2 = stats.norm.pdf(draws_co2, beta, sigma)
pdf_co2
```

Each value is the height of the Normal(5, 4) density curve at the corresponding observation. It is **not** a probability (a continuous variable has probability 0 of taking any exact value), but it tells us how *likely* values in the neighbourhood of that observation are, relative to others. Values close to the mean of 5 tonnes get the highest density (the maximum is $\frac{1}{\sigma\sqrt{2\pi}} \approx 0.199$), while observations far in the tails (e.g. close to 0 or 11 tonnes) get densities close to 0.

- Use a built-in function inside `stats.norm` to obtain the values of the cumulative density function (cumulated probabilities) for `draws_co2`, and comment what these values mean.

```python
cdf_co2 = stats.norm.cdf(draws_co2, beta, sigma)
cdf_co2
```

Each value is the probability that a randomly drawn individual emits **less than or equal to** that observation: $F(y) = P(Y \leq y)$. For example, the first observation is 4.59 tonnes and its CDF is 0.42, meaning 42% of the population emits 4.59 tonnes per capita or less. An observation equal to the mean (5 tonnes) would have a CDF of exactly 0.5, and the values always lie between 0 and 1.

- Use a built-in function inside `stats.norm` to obtain the 95th quantile (or percentile) of the distribution of CO2 emissions per capita (conceptually, it's the inverse of the one above) and explain what is the 95th quantile / percentile.

```python
q95 = stats.norm.ppf(0.95, beta, sigma)
q95
```

The 95th quantile (or percentile) is the value below which 95% of the distribution lies: $P(Y \leq q_{0.95}) = 0.95$. Here it is about **8.29 tonnes** ($= 5 + 1.645 \times 2$): 95% of individuals emit less than 8.29 tonnes of CO2 per capita, and only 5% emit more. The Percent Point Function (PPF) is the inverse of the CDF: we give it a probability and it returns the value.

- Use a combination of the generated array of CO2 emissions per capita, the Cumulative Density Function, and the Percentage Point Function to re-obtain the array of CO2 emissions per capita (tip: the PPF is the inverse of the CDF).

```python
# CDF turns values into probabilities, PPF turns them back into values
draws_co2_back = stats.norm.ppf(stats.norm.cdf(draws_co2, beta, sigma), beta, sigma)

# check that we recover the original array
print(np.allclose(draws_co2_back, draws_co2))
draws_co2_back
```

## Exercise 3: The one-sample z-test (unknown mean and known variance) <a name="Exercise-3:-The-one-sample-z-test-(unknown-mean-and-known-variance)"></a>

- In the resources' notebook we have run a one-sample z-test on the mean with known standard deviation $\sigma=2$.
- Specifically, we tested if, based on our result, the population mean equals 5 tonnes per capita ($H_0$) or if it's different than 5 tonnes per capita ($H_1$). We were not able to reject the null hypothesis that the population mean equals 5 tonnes per capita.
- Here, use the code seen in the resources notebook to test if the population mean equals 0 tonnes per capita ($H_0$) or if it is different than 0 tonnes per capita.
- First, compute the z-statistic and comment on whether you accept or reject the null hypothesis based on the z-statistic.

```python
beta_hat = st.mean(draws_co2)
sd_true = 2
beta_null = 0
N = len(draws_co2)
sem_true = sd_true / np.sqrt(N)
z = (beta_hat - beta_null) / sem_true
z.round(6)
```

The z-statistic is about **25.3**, far outside the interval $[-1.96, 1.96]$ that contains 95% of the values of a standard normal. Under $H_0: \beta = 0$ it would be extremely unlikely to observe such a value, so we **reject the null hypothesis** at the 5% level: the population mean of CO2 emissions per capita is different from 0.

- Then, compute the p value and comment on whether you accept or reject the null hypothesis based on the p value.

```python
lower_area = st.NormalDist().cdf(-abs(z))
upper_area = lower_area
p = lower_area + upper_area
p
```

Your answer here: The p-value is essentially **0** (numerically it rounds to 0.0, far below $10^{-100}$). Since $p < 0.05$, we **reject the null hypothesis** that the population mean equals 0 tonnes per capita.

- Then, as explained in the resources notebook, write down the equation of the 95% confidence interval under $H_0$ with $\beta_0=0$ tonnes per capita and known standard deviation $\sigma=2$ tonnes per capita.

Your answer here: With known standard deviation $\sigma = 2$ and $N = 100$, the standard error is $SEM = \frac{\sigma}{\sqrt{N}} = \frac{2}{10} = 0.2$, so the 95% confidence interval is:
$$CI_{95} = \hat{\beta} \pm 1.96 \times \frac{\sigma}{\sqrt{N}} = \hat{\beta} \pm 1.96 \times 0.2$$
We reject $H_0: \beta = \beta_0 = 0$ if $0$ lies outside this interval. (Equivalently, on the raw scale, we reject if $\hat{\beta}$ falls outside $\beta_0 \pm 1.96 \times 0.2 = [-0.392, 0.392]$.)

- Finally, use python to compute this confidence interval and use it to either accept or reject the null hypothesis.

```python
z_crit = stats.norm.ppf(0.975)
ci_lower = beta_hat - z_crit * sem_true
ci_upper = beta_hat + z_crit * sem_true
print(f"95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")
print("0 in CI:", ci_lower <= beta_null <= ci_upper)
```

The 95% confidence interval is about **[4.68, 5.46]**. It does not contain 0, so we **reject the null hypothesis** $H_0: \beta = 0$, consistent with the z-statistic and the p-value.

## Exercise 4: The one-sample t-test (unknown mean and variance) <a name="Exercise-4:-The-one-sample-t-test-(unknown-mean-and-variance)"></a>

- If we do not know the variance, our test statistic will be as follows:
$$z_{\hat{\beta}} = \frac{\hat{\beta}-\beta_0}{\hat{\sigma}/\sqrt{N}}$$
<br>
- If this estimate has been constructed from $N$ observations, then the sampling distribution turns into a $t$-distribution with $N-1$ **_degrees of freedom_** (df). The $t$ distribution is very similar to the normal distribution, but has "heavier" tails.
- Similarly to what has been shown in class for the one-sample z test, run a t test with the null hypothesis $\beta = 5$. Specifically, compute the t statistic and comment on whether you accept or reject the null hypothesis.

```python
beta_null = 5
N = len(draws_co2)
sd_hat = st.stdev(draws_co2)          # sample standard deviation (divides by N-1)
sem_hat = sd_hat / np.sqrt(N)
t = (beta_hat - beta_null) / sem_hat
t.round(6)
```

The t-statistic is about **0.32**. With $N-1 = 99$ degrees of freedom, the 5% critical values of the t-distribution are about $\pm 1.98$. Since $-1.98 < 0.32 < 1.98$, the statistic is not in the critical region, so we **cannot reject the null hypothesis** that the population mean equals 5 tonnes per capita.

- Use this test statistic to obtain the p value and comment on whether you accept or reject the null hypothesis based on the p value.

```python
lower_area = stats.t.cdf(-abs(t), N-1)
upper_area = lower_area
p = lower_area + upper_area
round(p, 6)
```

The p-value is about **0.75**, much larger than 0.05, so we **cannot reject the null hypothesis**: the sample mean (5.07 tonnes) is not statistically different from 5.

- Use a built-in function in `scipy.stats` to check that you computed correctly t stat and p value.

```python
stats.ttest_1samp(draws_co2, popmean = 5)
```

- Now, use a built-in function inside `stats.t` to get the the critical values for the areas with cumulative probability $0.025$ and $0.975$ of your t-student distribution.

```python
t_crit = stats.t.ppf([0.025, 0.975], N-1)
t_crit
```

- Use these critical values and the estimate for the standard error to write down the equation of the 95% confidence interval under $H_0$ with $\beta_0=0$ tonnes per capita and unknown standard deviation.

Your answer here: With unknown standard deviation, we replace $\sigma$ with its estimate $\hat{\sigma}$ and the normal critical value 1.96 with the t critical value $t_{99,\,0.975} \approx 1.984$:
$$CI_{95} = \hat{\beta} \pm t_{N-1,\,0.975} \times \frac{\hat{\sigma}}{\sqrt{N}} = \hat{\beta} \pm 1.984 \times \frac{\hat{\sigma}}{\sqrt{100}}$$
The interval does not depend on $\beta_0$: we reject $H_0: \beta = \beta_0 = 0$ if 0 lies outside it (and we can use the same interval to test $\beta_0 = 5$).

- Compute this confidence interval via python and use it to either accept or reject the null hypothesis.

```python
ci_lower = beta_hat + t_crit[0] * sem_hat
ci_upper = beta_hat + t_crit[1] * sem_hat
print(f"95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")
print("0 in CI:", ci_lower <= 0 <= ci_upper)
print("5 in CI:", ci_lower <= 5 <= ci_upper)

# The interval is about [4.65, 5.48]. It does not contain 0, so we reject H0: beta = 0.
# It contains 5, so (consistent with the t-test above) we cannot reject H0: beta = 5.
# It is slightly wider than the z-test interval [4.68, 5.46] because we estimate sigma.
```

- Use a built-in function in `scipy.stats` to check that you computed correctly the confidence interval.

```python
stats.t.interval(0.95, df = N-1, loc = beta_hat, scale = stats.sem(draws_co2))
```
